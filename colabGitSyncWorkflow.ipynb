```
# 1. USER CONFIGURATION (Change these!)

GITHUB_USER = "swayam-shreee"
REPO_NAME = "SolarFlareForcasting"
EMAIL = "@swayam969@gmail.com"
DRIVE_PROJECT_PATH = "/content/drive/MyDrive/solar_flare_forecasting"
```

```
# 2. MOUNT DRIVE & ACCESS TOKEN

from google.colab import drive, userdata
import os

# Mount your Google Drive
if not os.path.exists('/content/drive'):
    drive.mount('/content/drive')

# Safely retrieve your token from the Colab vault
try:
    GITHUB_TOKEN = userdata.get('GITHUB_TOKEN')
except Exception:
    raise ValueError("❌ GITHUB_TOKEN not found in Colab Secrets! Please add it via the Key icon.")

# Navigate to your project folder on Google Drive
%cd {DRIVE_PROJECT_PATH}
```

```
# 3. INITIALIZE GIT & CREDENTIALS

# Initialize git if it hasn't been done yet in this folder
if not os.path.exists('.git'):
    print("Initializing a fresh Git repository...")
    !git init
else:
    print("Git repository already initialized.")

# Set up global user identification
!git config --global user.name "{GITHUB_USER}"
!git config --global user.email "{EMAIL}"

# Securely construct the URL using the Token variable (hiding it from the cell output)
secure_remote_url = f"https://{GITHUB_USER}:{GITHUB_TOKEN}@github.com/{GITHUB_USER}/{REPO_NAME}.git"

# Reset the remote origin to ensure it uses the secure token url smoothly
!git remote remove origin 2>/dev/null
!git remote add origin {secure_remote_url}
print("✅ Secure Git configuration complete.")
```
